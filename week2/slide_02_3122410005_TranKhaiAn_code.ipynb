## Python List Methods

```python
x = [60, 70, 80, 90]
print("Ban đầu:", x)

# append(): thêm 1 phần tử vào cuối list
x.append(100)
print("append(100):", x)

# clear(): xóa toàn bộ phần tử trong list
y = x.copy()   # tạo bản sao trước
y.clear()
print("clear():", y)

# copy(): trả về một bản sao của list
y = x.copy()
print("copy():", y)

# count(): đếm số lần xuất hiện của 1 giá trị
print("count(70):", x.count(70))

# extend(): nối list khác vào list hiện tại
x.extend([110, 120])
print("extend([110, 120]):", x)

# index(): trả về vị trí đầu tiên tìm thấy của 1 giá trị
print("index(80):", x.index(80))

# insert(): thêm phần tử vào vị trí chỉ định
x.insert(2, 75)   # thêm 75 vào vị trí số 2
print("insert(2, 75):", x)

# pop(): xóa phần tử ở vị trí chỉ định (mặc định là cuối)
x.pop()  # xóa phần tử cuối cùng
print("pop():", x)
x.pop(2)  # xóa phần tử ở vị trí 2
print("pop(2):", x)

# remove(): xóa phần tử theo giá trị
x.remove(70)
print("remove(70):", x)

# reverse(): đảo ngược list
x.reverse()
print("reverse():", x)

# sort(): sắp xếp list tăng dần
x.sort()
print("sort():", x)
```

```python
# Python if elif else
scores = [90, 75, 89, 69, 55]
grades = []
for score in scores:
  if score >=90:
    grade = "A"
  elif score >= 80:
    grade = "B"
  elif score >= 70:
    grade = "C"
  elif score >= 60:
    grade = "D"
  else:
    grade = "E"

  grades.append(grade)

for i in range(len(scores)):
  print(f"Score: {scores[i]} -> Grade: {grades[i]}")
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def myfunc(self):
    print("Hello my name is " + self.name)

p1 = Person("Alan", 20)
p1.myfunc()
print(p1.name)
print(p1.age)
```

```python
class Vehicle:
    def __init__(self, name, kind="car", color="", value=100.00):
        self.name = name
        self.kind = kind
        self.color = color
        self.value = value

    def description(self):
        return f"{self.name} is a {self.color} {self.kind} worth ${self.value:,.2f}"

car1 = Vehicle("Fer", "convertible", "red", 60000.00)
car2 = Vehicle("Jump", "van", "blue", 10000.00)

print("Car 1:", car1.description())
print("Car 2:", car2.description())
```

## File Handling

• The key function for working with files in Python is the open() function.

• The open() function takes two parameters; filename, and mode.

• There are four different methos (modes) for opening a file:

    • "r" - Read - Default value. Opens a file for reading, error if the file does not exist.
    • "a" - Append - Opens a file for appending, creates the file if it does not exist.
    • "w" - Write - Opens a file for writing, creates the file if it does not exist.
    • "x" - Create - Creates the specified file, returns an error if the file exists.

```python
filename = "myfile.txt"

# 1. "w" - Write: tạo file mới (hoặc ghi đè nếu file đã tồn tại)
with open(filename, "w") as f:
    f.write("Hello World\n")
    f.write("This is a test file.\n")
print("Đã ghi file với chế độ 'w'.")

# 2. "r" - Read: đọc nội dung file
with open(filename, "r") as f:
    text = f.read()
print("Nội dung sau khi đọc với 'r':")
print(text)

# 3. "a" - Append: thêm nội dung vào cuối file
with open(filename, "a") as f:
    f.write("Appending a new line.\n")
print("Đã ghi thêm nội dung với chế độ 'a'.")

# Đọc lại để kiểm tra
with open(filename, "r") as f:
    print("Nội dung sau khi append:")
    print(f.read())

# 4. "x" - Create: tạo file mới, báo lỗi nếu đã tồn tại
try:
    with open("newfile.txt", "x") as f:
        f.write("This is a new file created with 'x' mode.\n")
    print("Đã tạo file 'newfile.txt' với chế độ 'x'.")
except FileExistsError:
    print("File 'newfile.txt' đã tồn tại, không thể tạo bằng 'x'.")
```

```python
#Python try: except: else: finally:
try:
  price = float(input("Enter the price of the stock (e.g. 10):"))
  shares = int(input("Enter the number of shares (e.g. 2):"))
  total = price * shares
except Exception as e:
  print("Exception error:", str(e))
else:
  print("The total value of the shares is:", total)
finally:
  print("Thank you.")
```
